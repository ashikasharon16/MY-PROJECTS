```python
import pandas as pd
import numpy as np

# Load dataset
df = pd.read_csv('/content/diabetes_dataset.csv')

print("Dataset loaded successfully!")
print("=" * 60)

print("Number of rows    :", df.shape[0])
print("Number of columns :", df.shape[1])

print("\nColumn Names:")
print(df.columns.tolist())

print("\nFirst 5 Rows:")
display(df.head())

print("\nDataset Information:")
df.info()
```

```python
# ============================================
# CELL 2: DATA QUALITY CHECK
# ============================================

print("DATA QUALITY CHECK")
print("=" * 60)

# 1. Missing values
print("\n1. Missing Values")
print("-" * 40)

missing_values = df.isnull().sum()
missing_percent = (df.isnull().sum() / len(df)) * 100

missing_summary = pd.DataFrame({
    'Missing Values': missing_values,
    'Missing Percentage': missing_percent
})

display(missing_summary)

# 2. Duplicate rows
print("\n2. Duplicate Rows")
print("-" * 40)

duplicate_count = df.duplicated().sum()
print("Number of duplicate rows:", duplicate_count)

# 3. Target distribution
print("\n3. Target Variable Distribution")
print("-" * 40)

print(df['diabetes'].value_counts())

print("\nTarget Percentage:")
print(df['diabetes'].value_counts(normalize=True) * 100)

# 4. Unique values for categorical columns
print("\n4. Categorical Variables")
print("-" * 40)

print("Gender values:")
print(df['gender'].value_counts())

print("\nSmoking history values:")
print(df['smoking_history'].value_counts())

print("\nLocation values:")
print(df['location'].nunique(), "unique locations")

print(df['location'].value_counts().head(10))

# 5. Numerical summary
print("\n5. Numerical Feature Summary")
print("-" * 40)

display(df.describe().T)
```

```python
# ============================================
# CELL 3: DATA CLEANING + TARGET VISUALIZATION
# ============================================

import matplotlib.pyplot as plt
import seaborn as sns

# -----------------------------
# 1. Remove duplicate rows
# -----------------------------

before = len(df)

df = df.drop_duplicates().reset_index(drop=True)

after = len(df)

print("Duplicate Removal")
print("=" * 50)
print("Rows before removing duplicates :", before)
print("Rows after removing duplicates  :", after)
print("Duplicates removed              :", before - after)


# -----------------------------
# 2. Target distribution
# -----------------------------

print("\nTarget Distribution")
print("=" * 50)

target_counts = df['diabetes'].value_counts()

print(target_counts)

print("\nPercentage:")
print((target_counts / len(df) * 100).round(2))


# -----------------------------
# 3. Target distribution plot
# -----------------------------

plt.figure(figsize=(7,5))

sns.countplot(
    data=df,
    x='diabetes'
)

plt.title("Diabetes Class Distribution")
plt.xlabel("Diabetes (0 = No, 1 = Yes)")
plt.ylabel("Number of Patients")

plt.show()


# -----------------------------
# 4. Numerical feature distributions
# -----------------------------

numerical_features = [
    'age',
    'bmi',
    'hbA1c_level',
    'blood_glucose_level'
]

fig, axes = plt.subplots(2, 2, figsize=(14,10))

for ax, feature in zip(axes.ravel(), numerical_features):

    sns.histplot(
        data=df,
        x=feature,
        hue='diabetes',
        kde=True,
        ax=ax
    )

    ax.set_title(f"{feature} Distribution by Diabetes Status")

plt.tight_layout()
plt.show()
```

```python
# CELL 4 — Correlation Analysis

import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Select important numerical/clinical features
clinical_features = [
    'age',
    'bmi',
    'hbA1c_level',
    'blood_glucose_level',
    'hypertension',
    'heart_disease',
    'diabetes'
]

# Correlation matrix
corr_matrix = df[clinical_features].corr()

print("Correlation Matrix")
print("=" * 60)
print(corr_matrix.round(3))

# Heatmap
plt.figure(figsize=(10, 7))

sns.heatmap(
    corr_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    center=0,
    linewidths=0.5
)

plt.title("Correlation Between Clinical Features and Diabetes")
plt.tight_layout()
plt.show()
```

```python
# CELL 5 — Compare Clinical Features by Diabetes Status

features = [
    'age',
    'bmi',
    'hbA1c_level',
    'blood_glucose_level',
    'hypertension',
    'heart_disease'
]

# Calculate mean values for diabetic and non-diabetic groups
group_means = df.groupby('diabetes')[features].mean().round(2)

print("Mean Clinical Feature Values by Diabetes Status")
print("=" * 70)
print(group_means)

# Visual comparison
fig, axes = plt.subplots(2, 3, figsize=(15, 9))

for ax, feature in zip(axes.flat, features):
    sns.boxplot(
        data=df,
        x='diabetes',
        y=feature,
        ax=ax
    )
    ax.set_title(f'{feature} vs Diabetes')
    ax.set_xlabel('Diabetes (0 = No, 1 = Yes)')
    ax.set_ylabel(feature)

plt.tight_layout()
plt.show()
```

```python
# CELL 6 — Categorical Feature Analysis

categorical_features = [
    'gender',
    'smoking_history',
    'hypertension',
    'heart_disease'
]

for feature in categorical_features:

    print("\n" + "=" * 70)
    print(f"{feature.upper()} vs DIABETES")
    print("=" * 70)

    # Count table
    count_table = pd.crosstab(
        df[feature],
        df['diabetes']
    )

    print("\nCount:")
    print(count_table)

    # Percentage within each feature category
    percentage_table = pd.crosstab(
        df[feature],
        df['diabetes'],
        normalize='index'
    ) * 100

    print("\nPercentage:")
    print(percentage_table.round(2))
```

```python
# CELL 7 — Combined Clinical Risk Profile Analysis

# Create clinical risk profile groups
df['clinical_risk_count'] = (
    (df['hypertension'] == 1).astype(int) +
    (df['heart_disease'] == 1).astype(int) +
    (df['hbA1c_level'] >= 6.5).astype(int) +
    (df['blood_glucose_level'] >= 126).astype(int) +
    (df['bmi'] >= 25).astype(int)
)

# Diabetes rate for each risk-profile count
risk_profile = (
    df.groupby('clinical_risk_count')['diabetes']
      .agg(['count', 'sum', 'mean'])
      .reset_index()
)

risk_profile['diabetes_rate_%'] = (
    risk_profile['mean'] * 100
).round(2)

risk_profile = risk_profile.drop(columns=['mean'])

print("Combined Clinical Risk Profile")
print("=" * 70)
print(risk_profile)

# Plot diabetes rate by number of risk factors
plt.figure(figsize=(10, 6))

sns.barplot(
    data=risk_profile,
    x='clinical_risk_count',
    y='diabetes_rate_%'
)

plt.xlabel("Number of Clinical Risk Factors")
plt.ylabel("Diabetes Rate (%)")
plt.title("Diabetes Rate by Combined Clinical Risk Profile")
plt.tight_layout()
plt.show()
```

```python
# CELL 8 — Prepare Features and Target

# Features to be used for modelling
features = [
    'year',
    'gender',
    'age',
    'location',
    'race:AfricanAmerican',
    'race:Asian',
    'race:Caucasian',
    'race:Hispanic',
    'race:Other',
    'hypertension',
    'heart_disease',
    'smoking_history',
    'bmi',
    'hbA1c_level',
    'blood_glucose_level'
]

X = df[features].copy()
y = df['diabetes'].copy()

print("Feature matrix shape :", X.shape)
print("Target shape         :", y.shape)

print("\nCategorical columns:")
print(X.select_dtypes(include='object').columns.tolist())

print("\nNumerical columns:")
print(X.select_dtypes(exclude='object').columns.tolist())
```

```python
# CELL 9 — Train/Test Split

from sklearn.model_selection import train_test_split

# Split the data while preserving the diabetes class ratio
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

print("Training set")
print("=" * 50)
print("X_train shape:", X_train.shape)
print("y_train shape:", y_train.shape)

print("\nTesting set")
print("=" * 50)
print("X_test shape :", X_test.shape)
print("y_test shape :", y_test.shape)

print("\nTraining target distribution:")
print(y_train.value_counts())
print((y_train.value_counts(normalize=True) * 100).round(2))

print("\nTesting target distribution:")
print(y_test.value_counts())
print((y_test.value_counts(normalize=True) * 100).round(2))
```

```python
# CELL 10 — Encode Categorical Features

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder

# Identify categorical and numerical columns
categorical_columns = [
    'gender',
    'location',
    'smoking_history'
]

numerical_columns = [
    col for col in X_train.columns
    if col not in categorical_columns
]

# Create preprocessing pipeline
preprocessor = ColumnTransformer(
    transformers=[
        (
            'categorical',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            ),
            categorical_columns
        ),
        (
            'numerical',
            'passthrough',
            numerical_columns
        )
    ]
)

# Fit only on training data and transform both datasets
X_train_processed = preprocessor.fit_transform(X_train)
X_test_processed = preprocessor.transform(X_test)

print("Original training shape :", X_train.shape)
print("Processed training shape:", X_train_processed.shape)

print("Original testing shape  :", X_test.shape)
print("Processed testing shape :", X_test_processed.shape)
```

```python
# CELL 11 — Baseline Logistic Regression

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix,
    classification_report
)

# Train Logistic Regression with class balancing
logistic_model = LogisticRegression(
    class_weight='balanced',
    max_iter=1000,
    random_state=42
)

logistic_model.fit(X_train_processed, y_train)

# Predictions
y_pred = logistic_model.predict(X_test_processed)
y_prob = logistic_model.predict_proba(X_test_processed)[:, 1]

# Evaluation
print("LOGISTIC REGRESSION — BASELINE")
print("=" * 60)

print("Accuracy  :", round(accuracy_score(y_test, y_pred), 4))
print("Precision :", round(precision_score(y_test, y_pred), 4))
print("Recall    :", round(recall_score(y_test, y_pred), 4))
print("F1 Score  :", round(f1_score(y_test, y_pred), 4))
print("ROC-AUC   :", round(roc_auc_score(y_test, y_prob), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred))

print("\nClassification Report:")
print(classification_report(y_test, y_pred))
```

```python
# CELL 12 — Scaled Logistic Regression

from sklearn.preprocessing import StandardScaler
from sklearn.pipeline import Pipeline
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Create scaled Logistic Regression pipeline
scaled_logistic_model = Pipeline([
    ('scaler', StandardScaler()),
    ('model', LogisticRegression(
        class_weight='balanced',
        max_iter=2000,
        random_state=42
    ))
])

# Train
scaled_logistic_model.fit(X_train_processed, y_train)

# Predictions
y_pred_scaled = scaled_logistic_model.predict(X_test_processed)
y_prob_scaled = scaled_logistic_model.predict_proba(X_test_processed)[:, 1]

# Evaluation
print("SCALED LOGISTIC REGRESSION")
print("=" * 60)

print("Accuracy  :", round(accuracy_score(y_test, y_pred_scaled), 4))
print("Precision :", round(precision_score(y_test, y_pred_scaled), 4))
print("Recall    :", round(recall_score(y_test, y_pred_scaled), 4))
print("F1 Score  :", round(f1_score(y_test, y_pred_scaled), 4))
print("ROC-AUC   :", round(roc_auc_score(y_test, y_prob_scaled), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_scaled))
```

```python
# CELL 13 — Decision Tree

from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Train Decision Tree
decision_tree_model = DecisionTreeClassifier(
    class_weight='balanced',
    max_depth=8,
    min_samples_split=20,
    random_state=42
)

decision_tree_model.fit(X_train_processed, y_train)

# Predictions
y_pred_dt = decision_tree_model.predict(X_test_processed)
y_prob_dt = decision_tree_model.predict_proba(X_test_processed)[:, 1]

# Evaluation
print("DECISION TREE")
print("=" * 60)

print("Accuracy  :", round(accuracy_score(y_test, y_pred_dt), 4))
print("Precision :", round(precision_score(y_test, y_pred_dt), 4))
print("Recall    :", round(recall_score(y_test, y_pred_dt), 4))
print("F1 Score  :", round(f1_score(y_test, y_pred_dt), 4))
print("ROC-AUC   :", round(roc_auc_score(y_test, y_prob_dt), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_dt))
```

```python
# CELL 14 — Random Forest

from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Train Random Forest
random_forest_model = RandomForestClassifier(
    n_estimators=200,
    max_depth=12,
    min_samples_split=10,
    class_weight='balanced',
    random_state=42,
    n_jobs=-1
)

random_forest_model.fit(X_train_processed, y_train)

# Predictions
y_pred_rf = random_forest_model.predict(X_test_processed)
y_prob_rf = random_forest_model.predict_proba(X_test_processed)[:, 1]

# Evaluation
print("RANDOM FOREST")
print("=" * 60)

print("Accuracy  :", round(accuracy_score(y_test, y_pred_rf), 4))
print("Precision :", round(precision_score(y_test, y_pred_rf), 4))
print("Recall    :", round(recall_score(y_test, y_pred_rf), 4))
print("F1 Score  :", round(f1_score(y_test, y_pred_rf), 4))
print("ROC-AUC   :", round(roc_auc_score(y_test, y_prob_rf), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_rf))
```

```python
# CELL 15 — XGBoost

from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# Calculate class imbalance ratio
negative_count = (y_train == 0).sum()
positive_count = (y_train == 1).sum()

scale_pos_weight = negative_count / positive_count

print("Scale Pos Weight:", round(scale_pos_weight, 2))

# Train XGBoost
xgb_model = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

xgb_model.fit(X_train_processed, y_train)

# Predictions
y_pred_xgb = xgb_model.predict(X_test_processed)
y_prob_xgb = xgb_model.predict_proba(X_test_processed)[:, 1]

# Evaluation
print("\nXGBOOST")
print("=" * 60)

print("Accuracy  :", round(accuracy_score(y_test, y_pred_xgb), 4))
print("Precision :", round(precision_score(y_test, y_pred_xgb), 4))
print("Recall    :", round(recall_score(y_test, y_pred_xgb), 4))
print("F1 Score  :", round(f1_score(y_test, y_pred_xgb), 4))
print("ROC-AUC   :", round(roc_auc_score(y_test, y_prob_xgb), 4))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test, y_pred_xgb))
```

```python
# ============================================================
# CLINICAL RISK SCORE FEATURE
# ============================================================

import pandas as pd

# Create a copy so the original dataframe remains unchanged
df_enhanced = df.copy()

# Composite Clinical Risk Score
df_enhanced['clinical_risk_score'] = (
    (df_enhanced['hypertension'] == 1).astype(int) +
    (df_enhanced['heart_disease'] == 1).astype(int) +
    (df_enhanced['hbA1c_level'] >= 6.5).astype(int) +
    (df_enhanced['blood_glucose_level'] >= 126).astype(int) +
    (df_enhanced['bmi'] >= 25).astype(int)
)

print("=" * 65)
print("COMPOSITE CLINICAL RISK SCORE")
print("=" * 65)

print("\nScore distribution:")
print(df_enhanced['clinical_risk_score'].value_counts().sort_index())

print("\nDiabetes rate by clinical risk score:")
risk_summary = (
    df_enhanced.groupby('clinical_risk_score')['diabetes']
    .agg(['count', 'mean'])
    .reset_index()
)

risk_summary['diabetes_rate_%'] = risk_summary['mean'] * 100
risk_summary = risk_summary.drop(columns=['mean'])

display(risk_summary)

print("\nRisk score range:",
      df_enhanced['clinical_risk_score'].min(),
      "to",
      df_enhanced['clinical_risk_score'].max())
```

```python
# ============================================================
# ENHANCED XGBOOST WITH CLINICAL RISK SCORE
# ============================================================

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from xgboost import XGBClassifier
from sklearn.metrics import (
    accuracy_score,
    precision_score,
    recall_score,
    f1_score,
    roc_auc_score,
    confusion_matrix
)

# ------------------------------------------------------------
# 1. Features + engineered Clinical Risk Score
# ------------------------------------------------------------

enhanced_features = [
    'year',
    'gender',
    'age',
    'location',
    'race:AfricanAmerican',
    'race:Asian',
    'race:Caucasian',
    'race:Hispanic',
    'race:Other',
    'hypertension',
    'heart_disease',
    'smoking_history',
    'bmi',
    'hbA1c_level',
    'blood_glucose_level',
    'clinical_risk_score'
]

X_enhanced = df_enhanced[enhanced_features]
y_enhanced = df_enhanced['diabetes']

# ------------------------------------------------------------
# 2. Train-test split
# ------------------------------------------------------------

X_train_enh, X_test_enh, y_train_enh, y_test_enh = train_test_split(
    X_enhanced,
    y_enhanced,
    test_size=0.20,
    stratify=y_enhanced,
    random_state=42
)

# ------------------------------------------------------------
# 3. Preprocessing
# ------------------------------------------------------------

categorical_enhanced = [
    'gender',
    'location',
    'smoking_history'
]

numerical_enhanced = [
    'year',
    'age',
    'race:AfricanAmerican',
    'race:Asian',
    'race:Caucasian',
    'race:Hispanic',
    'race:Other',
    'hypertension',
    'heart_disease',
    'bmi',
    'hbA1c_level',
    'blood_glucose_level',
    'clinical_risk_score'
]

preprocessor_enhanced = ColumnTransformer(
    transformers=[
        (
            'cat',
            OneHotEncoder(
                handle_unknown='ignore',
                sparse_output=False
            ),
            categorical_enhanced
        ),
        (
            'num',
            'passthrough',
            numerical_enhanced
        )
    ]
)

X_train_enhanced_processed = preprocessor_enhanced.fit_transform(X_train_enh)
X_test_enhanced_processed = preprocessor_enhanced.transform(X_test_enh)

print("Enhanced training shape:", X_train_enhanced_processed.shape)
print("Enhanced testing shape :", X_test_enhanced_processed.shape)

# ------------------------------------------------------------
# 4. Class imbalance
# ------------------------------------------------------------

negative_count = (y_train_enh == 0).sum()
positive_count = (y_train_enh == 1).sum()

scale_pos_weight_enhanced = negative_count / positive_count

print("Scale Pos Weight:", round(scale_pos_weight_enhanced, 2))

# ------------------------------------------------------------
# 5. Enhanced XGBoost
# ------------------------------------------------------------

xgb_enhanced = XGBClassifier(
    n_estimators=300,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=scale_pos_weight_enhanced,
    eval_metric='logloss',
    random_state=42,
    n_jobs=-1
)

xgb_enhanced.fit(
    X_train_enhanced_processed,
    y_train_enh
)

# ------------------------------------------------------------
# 6. Predictions
# ------------------------------------------------------------

y_pred_enhanced = xgb_enhanced.predict(
    X_test_enhanced_processed
)

y_prob_enhanced = xgb_enhanced.predict_proba(
    X_test_enhanced_processed
)[:, 1]

# ------------------------------------------------------------
# 7. Evaluation
# ------------------------------------------------------------

print("\n" + "=" * 60)
print("ENHANCED XGBOOST RESULTS")
print("=" * 60)

print("Accuracy  :", round(
    accuracy_score(y_test_enh, y_pred_enhanced), 4
))

print("Precision :", round(
    precision_score(y_test_enh, y_pred_enhanced), 4
))

print("Recall    :", round(
    recall_score(y_test_enh, y_pred_enhanced), 4
))

print("F1 Score  :", round(
    f1_score(y_test_enh, y_pred_enhanced), 4
))

print("ROC-AUC   :", round(
    roc_auc_score(y_test_enh, y_prob_enhanced), 4
))

print("\nConfusion Matrix:")
print(confusion_matrix(y_test_enh, y_pred_enhanced))
```

```python

# NEXT CELL: FEATURE ABLATION EXPERIMENT
# Compare model with all features vs without HbA1c and glucose

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score,
    f1_score, roc_auc_score
)
from xgboost import XGBClassifier

# Remove duplicate rows for this experiment
experiment_df = df.drop_duplicates().copy()

target = "diabetes"
X = experiment_df.drop(columns=[target])
y = experiment_df[target].astype(int)

# Same test set for both experiments
X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.20,
    random_state=42,
    stratify=y
)

categorical_cols = ["gender", "location", "smoking_history"]

def run_ablation_experiment(X_train_part, X_test_part, y_train, y_test, label):
    categorical = [c for c in categorical_cols if c in X_train_part.columns]
    numeric = [c for c in X_train_part.columns if c not in categorical]

    preprocessor = ColumnTransformer(
        transformers=[
            ("cat", OneHotEncoder(handle_unknown="ignore"), categorical),
            ("num", "passthrough", numeric)
        ]
    )

    X_train_ready = preprocessor.fit_transform(X_train_part)
    X_test_ready = preprocessor.transform(X_test_part)

    model = XGBClassifier(
        n_estimators=250,
        max_depth=6,
        learning_rate=0.05,
        subsample=0.8,
        colsample_bytree=0.8,
        scale_pos_weight=(y_train == 0).sum() / (y_train == 1).sum(),
        eval_metric="logloss",
        random_state=42,
        n_jobs=-1
    )

    model.fit(X_train_ready, y_train)

    probabilities = model.predict_proba(X_test_ready)[:, 1]
    predictions = (probabilities >= 0.5).astype(int)

    return {
        "Experiment": label,
        "Features after encoding": X_train_ready.shape[1],
        "Accuracy": round(accuracy_score(y_test, predictions), 4),
        "Precision": round(precision_score(y_test, predictions, zero_division=0), 4),
        "Recall": round(recall_score(y_test, predictions, zero_division=0), 4),
        "F1": round(f1_score(y_test, predictions, zero_division=0), 4),
        "ROC-AUC": round(roc_auc_score(y_test, probabilities), 4)
    }

# Experiment 1: all available features
full_result = run_ablation_experiment(
    X_train, X_test, y_train, y_test,
    "All features"
)

# Experiment 2: remove the two clinical measurements
removed_features = ["hbA1c_level", "blood_glucose_level"]

X_train_reduced = X_train.drop(columns=removed_features)
X_test_reduced = X_test.drop(columns=removed_features)

reduced_result = run_ablation_experiment(
    X_train_reduced, X_test_reduced, y_train, y_test,
    "Without HbA1c and glucose"
)

comparison_df = pd.DataFrame([full_result, reduced_result])

print("FEATURE ABLATION RESULTS")
display(comparison_df)

print("\nPerformance change after removing HbA1c and glucose:")
for metric in ["Accuracy", "Precision", "Recall", "F1", "ROC-AUC"]:
    change = reduced_result[metric] - full_result[metric]
    print(f"{metric}: {change:+.4f}")
```

```python

# NEXT CELL: PROBABILITY CALIBRATION AUDIT

import pandas as pd
import numpy as np

from sklearn.model_selection import train_test_split
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.metrics import roc_auc_score, brier_score_loss, log_loss
from xgboost import XGBClassifier

# Prepare dataset
cal_df = df.drop_duplicates().copy()
X = cal_df.drop(columns=["diabetes"])
y = cal_df["diabetes"].astype(int)

categorical_cols = ["gender", "location", "smoking_history"]
numeric_cols = [c for c in X.columns if c not in categorical_cols]

# Keep a held-out test portion; use training portion for this audit
X_train_full, X_test_holdout, y_train_full, y_test_holdout = train_test_split(
    X, y, test_size=0.20, random_state=42, stratify=y
)

# Split training portion into model-training and validation subsets
X_subtrain, X_val, y_subtrain, y_val = train_test_split(
    X_train_full, y_train_full,
    test_size=0.20, random_state=21, stratify=y_train_full
)

preprocessor = ColumnTransformer([
    ("cat", OneHotEncoder(handle_unknown="ignore"), categorical_cols),
    ("num", "passthrough", numeric_cols)
])

X_subtrain_ready = preprocessor.fit_transform(X_subtrain)
X_val_ready = preprocessor.transform(X_val)

model = XGBClassifier(
    n_estimators=250,
    max_depth=6,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    scale_pos_weight=(y_subtrain == 0).sum() / (y_subtrain == 1).sum(),
    eval_metric="logloss",
    random_state=42,
    n_jobs=-1
)

model.fit(X_subtrain_ready, y_subtrain)

# Predicted probabilities on validation data
val_probs = model.predict_proba(X_val_ready)[:, 1]

print("=" * 60)
print("VALIDATION PROBABILITY CALIBRATION AUDIT")
print("=" * 60)
print("Validation rows:", len(y_val))
print("ROC-AUC:", round(roc_auc_score(y_val, val_probs), 4))
print("Brier score (lower is better):", round(brier_score_loss(y_val, val_probs), 4))
print("Log loss (lower is better):", round(log_loss(y_val, val_probs), 4))

# Compare average predicted probability with observed positive rate
calibration_df = pd.DataFrame({
    "predicted_probability": val_probs,
    "actual_label": y_val.to_numpy()
})

calibration_df["probability_group"] = pd.cut(
    calibration_df["predicted_probability"],
    bins=np.linspace(0, 1, 11),
    include_lowest=True
)

calibration_table = calibration_df.groupby(
    "probability_group", observed=False
).agg(
    patient_count=("actual_label", "size"),
    mean_predicted_probability=("predicted_probability", "mean"),
    observed_positive_rate=("actual_label", "mean")
)

calibration_table["mean_predicted_probability"] = (
    calibration_table["mean_predicted_probability"].round(4)
)
calibration_table["observed_positive_rate"] = (
    calibration_table["observed_positive_rate"].round(4)
)

print("\nCalibration table:")
display(calibration_table)

print("\nNote: Empty probability bins have no observations.")
print("This is a validation analysis, not clinical validation.")
```

```python

# NEXT CELL: RAW vs CALIBRATED XGBOOST
# Validation data calibrates probabilities; held-out test evaluates them.

import numpy as np
import pandas as pd

from sklearn.linear_model import LogisticRegression
from sklearn.metrics import (
    roc_auc_score, brier_score_loss, log_loss
)
from scipy.special import logit, expit

# Use the split variables from the previous calibration cell.
# Fit preprocessing and XGBoost on the sub-training data only.
X_subtrain_ready = preprocessor.fit_transform(X_subtrain)
X_val_ready = preprocessor.transform(X_val)
X_test_ready = preprocessor.transform(X_test_holdout)

model.fit(X_subtrain_ready, y_subtrain)

# Raw predicted probabilities
val_raw = model.predict_proba(X_val_ready)[:, 1]
test_raw = model.predict_proba(X_test_ready)[:, 1]

# Platt-style sigmoid calibration fitted on validation predictions
eps = 1e-6
val_logits = logit(np.clip(val_raw, eps, 1 - eps)).reshape(-1, 1)
test_logits = logit(np.clip(test_raw, eps, 1 - eps)).reshape(-1, 1)

calibrator = LogisticRegression()
calibrator.fit(val_logits, y_val)

test_calibrated = calibrator.predict_proba(test_logits)[:, 1]

# Compare on the same untouched test set
results = []

for name, probs in [
    ("Raw XGBoost", test_raw),
    ("Sigmoid-calibrated XGBoost", test_calibrated)
]:
    results.append({
        "Model": name,
        "ROC-AUC": round(roc_auc_score(y_test_holdout, probs), 4),
        "Brier score": round(brier_score_loss(y_test_holdout, probs), 4),
        "Log loss": round(log_loss(y_test_holdout, probs), 4)
    })

print("TEST-SET CALIBRATION COMPARISON")
display(pd.DataFrame(results))

# Calibration table for calibrated test probabilities
test_calibration = pd.DataFrame({
    "probability": test_calibrated,
    "actual": np.asarray(y_test_holdout)
})

test_calibration["probability_group"] = pd.cut(
    test_calibration["probability"],
    bins=np.linspace(0, 1, 11),
    include_lowest=True
)

table = test_calibration.groupby(
    "probability_group", observed=False
).agg(
    patient_count=("actual", "size"),
    mean_predicted_probability=("probability", "mean"),
    observed_positive_rate=("actual", "mean")
)

print("\nCALIBRATED MODEL: TEST-SET CALIBRATION TABLE")
display(table.round(4))

print(
    "\nInterpretation guide: lower Brier score and log loss are better. "
    "ROC-AUC measures ranking, not calibration."
)
```

```python

# NEXT CELL: SUBGROUP PERFORMANCE & RELIABILITY AUDIT

import pandas as pd
import numpy as np
from sklearn.metrics import roc_auc_score, brier_score_loss, recall_score

# Match test rows with their true labels and predicted probabilities
subgroup_df = X_test_holdout[["gender", "age"]].copy()
subgroup_df["actual"] = np.asarray(y_test_holdout)
subgroup_df["raw_probability"] = np.asarray(test_raw)
subgroup_df["calibrated_probability"] = np.asarray(test_calibrated)

# Create age groups for analysis
subgroup_df["age_group"] = pd.cut(
    subgroup_df["age"],
    bins=[0, 18, 30, 45, 60, 120],
    labels=["0-18", "19-30", "31-45", "46-60", "60+"],
    include_lowest=True
)

def summarize_group(group, group_name, group_value):
    actual = group["actual"].astype(int)
    raw_prob = group["raw_probability"]
    calibrated_prob = group["calibrated_probability"]

    result = {
        "Group type": group_name,
        "Group": str(group_value),
        "Rows": len(group),
        "Actual positive rate (%)": round(actual.mean() * 100, 2),
        "Raw Brier score": round(brier_score_loss(actual, raw_prob), 4),
        "Calibrated Brier score": round(
            brier_score_loss(actual, calibrated_prob), 4
        ),
        "Calibrated recall at 0.5": round(
            recall_score(actual, calibrated_prob >= 0.5, zero_division=0), 4
        )
    }

    # AUC is undefined if the group contains only one target class
    if actual.nunique() == 2:
        result["Calibrated ROC-AUC"] = round(
            roc_auc_score(actual, calibrated_prob), 4
        )
    else:
        result["Calibrated ROC-AUC"] = np.nan

    return result

rows = []

for group_value, group in subgroup_df.groupby("gender", dropna=False):
    rows.append(summarize_group(group, "Gender", group_value))

for group_value, group in subgroup_df.groupby("age_group", observed=False, dropna=False):
    if len(group) > 0:
        rows.append(summarize_group(group, "Age group", group_value))

subgroup_results = pd.DataFrame(rows)

print("SUBGROUP PERFORMANCE AUDIT")
display(subgroup_results)

print("\nNote:")
print("- Small subgroups can have unstable metrics.")
print("- Differences do not establish bias or clinical validity by themselves.")
print("- This is an exploratory audit on the held-out test set.")
```
